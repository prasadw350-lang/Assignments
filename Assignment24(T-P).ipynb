Import Libraries

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

Classification Dataset

```python
filepath = "/content/drive/MyDrive/mushrooms.csv"
cls_df = pd.read_csv(filepath)

cls_df.head()
```

```python
cls_df.info()

cls_df.isnull().sum()
```

```python
le = LabelEncoder()
for col in cls_df.columns:
    cls_df[col] = le.fit_transform(cls_df[col])
```

```python
X_cls = cls_df.drop("class", axis=1)
y_cls = cls_df["class"]
```

```python
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42
)
```

```python
scaler_cls = StandardScaler()
X_train_cls = scaler_cls.fit_transform(X_train_cls)
X_test_cls = scaler_cls.transform(X_test_cls)
```

Regression Dataset

```python
filepath = "/content/drive/MyDrive/insurance.csv"
reg_df = pd.read_csv(filepath)

reg_df.head()
```

```python
reg_df.info()

reg_df.isnull().sum()
```

```python
reg_df = pd.get_dummies(reg_df, drop_first=True)

X_reg = reg_df.drop("charges", axis=1)
y_reg = reg_df["charges"]
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)
```

```python
num_cols = ["age", "bmi", "children"]

scaler_reg = StandardScaler()

X_train_reg[num_cols] = scaler_reg.fit_transform(X_train_reg[num_cols])

X_test_reg[num_cols] = scaler_reg.transform(X_test_reg[num_cols])
```

Clssification Algorithms

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

lr = LogisticRegression(max_iter=1000)

lr.fit(X_train_cls, y_train_cls)

lr_pred = lr.predict(X_test_cls)

lr_acc = accuracy_score(y_test_cls, lr_pred)

print("Accuracy:", lr_acc)
print(confusion_matrix(y_test_cls, lr_pred))
print(classification_report(y_test_cls, lr_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train_cls, y_train_cls)

dt_pred = dt.predict(X_test_cls)

dt_acc = accuracy_score(y_test_cls, dt_pred)

print("Accuracy:", dt_acc)
print(confusion_matrix(y_test_cls, dt_pred))
print(classification_report(y_test_cls, dt_pred))
```

```python
from sklearn.svm import SVC

svm = SVC()

svm.fit(X_train_cls, y_train_cls)

svm_pred = svm.predict(X_test_cls)

svm_acc = accuracy_score(y_test_cls, svm_pred)

print("Accuracy:", svm_acc)
print(confusion_matrix(y_test_cls, svm_pred))
print(classification_report(y_test_cls, svm_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train_cls, y_train_cls)

knn_pred = knn.predict(X_test_cls)

knn_acc = accuracy_score(y_test_cls, knn_pred)

print("Accuracy:", knn_acc)
print(confusion_matrix(y_test_cls, knn_pred))
print(classification_report(y_test_cls, knn_pred))
```

```python
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()

nb.fit(X_train_cls, y_train_cls)

nb_pred = nb.predict(X_test_cls)

nb_acc = accuracy_score(y_test_cls, nb_pred)

print("Accuracy:", nb_acc)
print(confusion_matrix(y_test_cls, nb_pred))
print(classification_report(y_test_cls, nb_pred))
```

```python
comparison = pd.DataFrame({
    "Algorithm": [
        "Logistic Regression",
        "Decision Tree",
        "Support Vector Machine",
        "K-Nearest Neighbors",
        "Naive Bayes"
    ],
    "Accuracy": [
        lr_acc,
        dt_acc,
        svm_acc,
        knn_acc,
        nb_acc
    ]
})

comparison
```

Regression Algorithms

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

lr_reg = LinearRegression()
lr_reg.fit(X_train_reg, y_train_reg)
lr_pred = lr_reg.predict(X_test_reg)
lr_r2 = r2_score(y_test_reg, lr_pred)

print("R2 Score:", lr_r2)
```

```python
from sklearn.tree import DecisionTreeRegressor

dt_reg = DecisionTreeRegressor(random_state=42)
dt_reg.fit(X_train_reg, y_train_reg)
dt_pred = dt_reg.predict(X_test_reg)
dt_r2 = r2_score(y_test_reg, dt_pred)

print("R2 Score:", dt_r2)
```

```python
from sklearn.svm import SVR

svr = SVR()
svr.fit(X_train_reg, y_train_reg)
svr_pred = svr.predict(X_test_reg)
svr_r2 = r2_score(y_test_reg, svr_pred)

print("R2 Score:", svr_r2)
```

```python
from sklearn.neighbors import KNeighborsRegressor

knn_reg = KNeighborsRegressor(n_neighbors=5)
knn_reg.fit(X_train_reg, y_train_reg)
knn_pred = knn_reg.predict(X_test_reg)
knn_r2 = r2_score(y_test_reg, knn_pred)

print("R2 Score:", knn_r2)
```

```python
comparison_reg = pd.DataFrame({
    "Algorithm": [
        "Linear Regression",
        "Decision Tree Regressor",
        "Support Vector Regressor",
        "K-Nearest Neighbors Regressor"
    ],
    "R2 Score": [
        lr_r2,
        dt_r2,
        svr_r2,
        knn_r2
    ]
})

comparison_reg
```

```python
best_reg = comparison_reg.loc[comparison_reg["R2 Score"].idxmax()]

print("Best Regression Model:", best_reg["Algorithm"])
print("R2 Score:", best_reg["R2 Score"])
```
