```python
import pandas as pd

from google.colab import files
uploaded = files.upload()

df = pd.read_csv("insurance.csv")
print(df.head(10))
```

```python
print("Shape:")
print(df.shape)

print("\nInfo:")
print(df.info())

print("\nDescription:")
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
numerical = df.select_dtypes(include=["int64","float64"])
categorical = df.select_dtypes(include=["object"])

print("Numerical Columns")
print(numerical.columns)

print("\nCategorical Columns")
print(categorical.columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
cols = ["age","bmi","children","expenses"]

for col in cols:
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in cols:
    sns.countplot(x=col,data=df)
    plt.title(col)
    plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm")

plt.show()
```

```python
print("Average Charges")
print(df["expenses"].mean())

print("\nMaximum Charges")
print(df["expenses"].max())

print("\nMinimum Charges")
print(df["expenses"].min())

print("\nAverage Charges by Smoker")
print(df.groupby("smoker")["expenses"].mean())
```

```python
sns.boxplot(x="smoker",y="expenses",hue="sex",data=df)

plt.title("Charges vs Smoker and Sex")
plt.show()
```

```python
print("1. Average age =", df["age"].mean())
print("Average BMI =", df["bmi"].mean())

print("\n2. Smokers have higher insurance charges than non-smokers.")

print("\n3. Region with highest customers:")
print(df["region"].value_counts().idxmax())

print("\n4. Observation:")
print("- There are 1338 records and 7 columns.")
print("- The Southeast region has the highest number of customers.")
print("- The dataset contains no missing values.")
```

```python
import pandas as pd
filepath = "/content/drive/MyDrive/anime.csv"
df = pd.read_csv(filepath)
df.head(10)
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nInfo:")
df.info()

print("\nDescription:")
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
num = df.select_dtypes(include=["int64","float64"])
cat = df.select_dtypes(include=["object"])

print("Numerical Columns")
print(num.columns)
print("\nCategorical Columns")
print(cat.columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cols=["score","rank","members","episodes"]

for col in cols:
    sns.histplot(df[col],kde=True)
    plt.title(col)
    plt.show()
```

```python
sns.countplot(x="type",data=df)

plt.title("Anime Type")

plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.show()
```

```python
print("Average Score:",df["score"].mean())

print("Maximum Members:",df["members"].max())

print("Minimum Members:",df["members"].min())

print(df.groupby("type")["score"].mean())
```

```python
sns.boxplot(x="type",
            y="score",
            data=df)

plt.show()
```

```python
print("\n1. Average Anime Score:")
print(df["score"].mean())

print("\n2. Average Number of Episodes:")
print(df["episodes"].mean())

print("\n3. Most Common Anime Type:")
print(df["type"].value_counts().idxmax())

print("\n4. Interesting Observations:")
print("- Total Records:", df.shape[0])
print("- Total Columns:", df.shape[1])
print("- Missing Values:")
print(df.isnull().sum())

print("\nHighest Score:", df["score"].max())
print("Lowest Score:", df["score"].min())
```
