Q1. (Data Preparation)

```python
import pandas as pd

df = pd.read_csv("cleaned_emotions.csv")
```

```python
X = df["cleaned_text"]
y = df["label"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

Q2. (Bag of Words – Basic)

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()

X_train_bow = cv.fit_transform(X_train)
X_test_bow = cv.transform(X_test)
```

```python
print(X_train_bow.shape)
print(X_test_bow.shape)
```

```python
print(cv.get_feature_names_out()[:20])
```

Q3. (Bag of Words + MultinomialNB)

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()

model.fit(X_train_bow, y_train)
```

```python
y_pred = model.predict(X_test_bow)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

Q4. (Understanding Vocabulary)

```python
print("Vocabulary Size:", len(cv.vocabulary_))
```

```python
print(cv.get_feature_names_out()[:15])
```

```python
sample = X_train.iloc[0]

vector = cv.transform([sample])

print(sample)
print(vector.toarray())
```

Q5. (N-grams with Bag of Words)

```python
from sklearn.feature_extraction.text import CountVectorizer

cv_ngram = CountVectorizer(ngram_range=(1, 2))

X_train_ngram = cv_ngram.fit_transform(X_train)
X_test_ngram = cv_ngram.transform(X_test)
```

```python
print(X_train_ngram.shape)
print(X_test_ngram.shape)
```

```python
print(cv_ngram.get_feature_names_out()[:20])
```

Q6. (N-grams + Model Training)

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

model_ngram = MultinomialNB()

model_ngram.fit(X_train_ngram, y_train)

y_pred_ngram = model_ngram.predict(X_test_ngram)

accuracy_ngram = accuracy_score(y_test, y_pred_ngram)

print("Bigram Accuracy:", accuracy_ngram)
```

```python
print("BoW Accuracy:", accuracy)
print("BoW + Bigram Accuracy:", accuracy_ngram)
```

Q7. (TF-IDF Vectorization)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
print(X_train_tfidf.shape)
print(X_test_tfidf.shape)
```

```python
print(tfidf.get_feature_names_out()[:15])
```

Q8. (TF-IDF + MultinomialNB)

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

model_tfidf = MultinomialNB()

model_tfidf.fit(X_train_tfidf, y_train)

y_pred_tfidf = model_tfidf.predict(X_test_tfidf)

accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("TF-IDF Accuracy:", accuracy_tfidf)
```

Q9. (Comparison of Vectorizers)

```python
comparison = pd.DataFrame({
    "Method": [
        "BoW Unigram",
        "BoW Unigram + Bigram",
        "TF-IDF"
    ],
    "Accuracy": [
        accuracy,
        accuracy_ngram,
        accuracy_tfidf
    ]
})

print(comparison)
```

The method with the highest accuracy performed best on the test dataset. The difference may be due to the way each vectorizer represents the text.

Q10. (Mini Project – Complete Vectorization Pipeline)

```python
import pandas as pd

df = pd.read_csv("cleaned_emotions.csv")
```

```python
X = df["cleaned_text"]
y = df["label"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

bow = CountVectorizer()

X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

model_bow = MultinomialNB()
model_tfidf = MultinomialNB()

model_bow.fit(X_train_bow, y_train)
model_tfidf.fit(X_train_tfidf, y_train)
```

```python
bow_pred = model_bow.predict(X_test_bow)
tfidf_pred = model_tfidf.predict(X_test_tfidf)

bow_acc = accuracy_score(y_test, bow_pred)
tfidf_acc = accuracy_score(y_test, tfidf_pred)

print("BoW Accuracy:", bow_acc)
print("TF-IDF Accuracy:", tfidf_acc)
```

```python
best_model = model_bow
best_vectorizer = bow
print("Best: BoW")
```

```python
import joblib

joblib.dump(best_model, "best_model.pkl")
joblib.dump(best_vectorizer, "best_vectorizer.pkl")

print("Saved successfully!")
```
