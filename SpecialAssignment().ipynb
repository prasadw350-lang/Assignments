```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["daily_playtime_hours"], kde=True)
plt.title("Daily Playtime Distribution")
plt.show()
```

```python
sns.countplot(x="gender", data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
plt.scatter(df["daily_playtime_hours"], df["addiction_score"])
plt.xlabel("Daily Playtime")
plt.ylabel("Addiction Score")
plt.title("Daily Playtime vs Addiction Score")
plt.show()
```

```python
sns.boxplot(x="addiction_binary",
            y="daily_playtime_hours",
            data=df)

plt.show()
```

```python
df["gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```
